# Solving the Max-Cut Problem Using QAOA

In this example, a quantum computer is used to solve the Max-Cut problem with QAOA (Quantum Approximate Optimization Algorithm).

This sample is designed to run on ABCI-Q. 
For instructions on setting up the Prefect environment on ABCI-Q and running workflows, please refer to the [Prefect Manual](https://g-quat-abciq.github.io/abciq-docs/ja/apps/prefect_install/) in the ABCI-Q User Guide.

This sample is based on the tutorial ["QAOA Explained: Solving Max-Cut Step by Step"](https://www.freequantumcomputing.com/ja/tutorials/qaoa-max-cut-tutorial/).
The original tutorial has been adapted to use quri-parts-oqtopus and modified to run as a Prefect workflow.

### 1. Import of Required Libraries

````python
from prefect import flow, task

import numpy as np
from scipy.optimize import minimize

from quri_parts.circuit import QuantumCircuit, Parameter
from quri_parts_oqtopus.backend import OqtopusConfig, OqtopusSamplingBackend
````

### 2. Creating the Quantum Circuit

A simple four-node graph is defined, and a QAOA circuit with two parameters is constructed using `create_qaoa_circuit`.

This circuit is executed by the `sample` function on a quantum computer (or simulator) running on ABCI-Q using quri-parts-oqtopus.

The parameters supplied to the QAOA circuit are optimized using the `minimize` function from the scipy library. In the `objective` function, the expected Max-Cut value is calculated from the samples obtained through circuit execution, and the parameter search is performed so as to maximize this value.

````python
# =====================================
#  Max Cut Edges
# =====================================
EDGES = [
    (0, 1),
    (1, 2),
    (2, 3),
    (3, 0),
]

N_QUBITS = 4

# =====================================
#  Max Cut QAOA Circuit
# =====================================
def create_qaoa_circuit(gamma: float, beta: float) -> QuantumCircuit:
    circuit = QuantumCircuit(N_QUBITS)

    for q in range(N_QUBITS):
        circuit.add_H_gate(q)

    for i, j in EDGES:
        circuit.add_CNOT_gate(i, j)
        circuit.add_RZ_gate(j, gamma)
        circuit.add_CNOT_gate(i, j)

    for q in range(N_QUBITS):
        circuit.add_RX_gate(q, 2.0 * beta)

    return circuit

def bitstring_from_int(value: int) -> str:
    return format(value, f"0{N_QUBITS}b")


def maxcut_score(bitstring: str) -> int:
    score = 0

    for i, j in EDGES:
        if bitstring[i] != bitstring[j]:
            score += 1

    return score

def compute_expected_cost(counts: dict[int, int]) -> float:
    total_shots = sum(counts.values())

    if total_shots == 0:
        raise ValueError("No samples were returned.")

    expectation = 0.0

    for state, count in counts.items():
        bitstring = bitstring_from_int(state)
        cost = maxcut_score(bitstring)
        expectation += cost * count

    return -(expectation / total_shots)

@task
def sample(circuit: QuantumCircuit, shots: int) -> dict[int, int]:
    backend = OqtopusSamplingBackend(OqtopusConfig.from_file("oqtopus-dev"))

    job = backend.sample(
        circuit,
        device_id="qiskit_sim",
        shots=shots,
    )

    return dict(job.result().counts)   

def objective(params):
    gamma, beta = params

    circuit = create_qaoa_circuit(gamma, beta)
    
    counts = sample(circuit, 2000)

    expectation = compute_expected_cost(counts)

    return expectation

@task
def optimize_electron():
    constraints = [
        {"type": "ineq", "fun": lambda x: x[0]},              # gamma >= 0
        {"type": "ineq", "fun": lambda x: np.pi - x[0]},      # gamma <= pi
        {"type": "ineq", "fun": lambda x: x[1]},              # beta >= 0
        {"type": "ineq", "fun": lambda x: np.pi / 2 - x[1]},  # beta <= pi/2
    ]

    initial_params = [0.7, 0.4]

    result = minimize(
        objective,
        initial_params,
        method="COBYLA",
        constraints=constraints,
        options={
            "maxiter": 50,
            "disp": True,
        },
    )

    return result
````

### 3. Final Sampling with the Optimal Parameters

In `final_sampling`, the optimal parameters passed as arguments are used to perform the final sampling on the quantum computer and obtain the result.

Here, the task is executed independently using the System-F plugin in Covalent.

````python
@task
def final_sampling(result):
    gamma_opt, beta_opt = result['x']

    print("\n=== Final Sampling ===")

    best_circuit = create_qaoa_circuit(
        gamma_opt,
        beta_opt,
    )

    counts = sample(best_circuit, 10000)

    return counts  
````

### 4. Defining and Executing the Covalent Workflow

The `workflow` is defined and then executed.

The `workflow` consists of two Covalent tasks: `optimize_electron`, which performs the optimization of the QAOA circuit parameters, and `final_sampling`, which performs the final sampling using the optimized parameters.

````python
@flow(log_prints=True)
def main():
    result = optimize_electron()
    counts = final_sampling(result)

    return counts

res = main()
print(res)
````

### 5. Displaying the Final Sampling Results

The sampling results are sorted and displayed in descending order of their counts, allowing the most probable and highest-quality solution to be identified.

Output example：
```
0101 : count=2721 cut=4
1010 : count=2627 cut=4
0011 : count=844 cut=2
1001 : count=830 cut=2
・・・
```

````python
def print_top_samples(counts: dict[int, int], top_k: int = 10) -> None:
    sorted_counts = sorted(
        counts.items(),
        key=lambda x: x[1],
        reverse=True,
    )

    for state, count in sorted_counts[:top_k]:
        bitstring = bitstring_from_int(state)
        score = maxcut_score(bitstring)

        print(
            f"{bitstring} : "
            f"count={count} "
            f"cut={score}"
        )

print_top_samples(res)
````
